```python
flight_df['Flight_day']=flight_df['fl_date'].str.split('-').str[2]
flight_df['Flight_Month']=flight_df['fl_date'].str.split('-').str[1]
flight_df['Flight_year']=flight_df['fl_date'].str.split('-').str[0]
```

```python
flight_df.drop('fl_date',axis=1,inplace=True)
```

```python
flight_df.head(5)
```

```python
flight_df.info()
```

```python
flight_df['Flight_day']=flight_df['Flight_day'].astype(int)
flight_df['Flight_Month']=flight_df['Flight_Month'].astype(int)
flight_df['Flight_year']=flight_df['Flight_year'].astype(int)
```

```python
flight_df.info()
```

```python
flight_df.isnull().sum()
```

```python
flight_df.drop('cancellation_code',axis=1,inplace=True)
```

```python
flight_df['dep_time']=flight_df['dep_time'].fillna(flight_df['dep_time'].mean(),inplace=True)
flight_df['dep_delay']=flight_df['dep_delay'].fillna(flight_df['dep_delay'].mean(),inplace=True)
flight_df['taxi_out']=flight_df['taxi_out'].fillna(flight_df['taxi_out'].mean(),inplace=True)
flight_df['wheels_off']=flight_df['wheels_off'].fillna(flight_df['wheels_off'].mean(),inplace=True)
flight_df['wheels_on']=flight_df['wheels_on'].fillna(flight_df['wheels_on'].mean(),inplace=True)
flight_df['taxi_in']=flight_df['taxi_in'].fillna(flight_df['taxi_in'].mean(),inplace=True)
flight_df['arr_time']=flight_df['arr_time'].fillna(flight_df['arr_time'].mean(),inplace=True)
flight_df['arr_delay']=flight_df['arr_delay'].fillna(flight_df['arr_delay'].mean(),inplace=True)
```

```python
flight_df.isnull().sum()
```

```python
flight_df = flight_df.dropna()
```

```python
flight_df.isnull().sum()
```

```python
flight_df
```

```python
flight_df['op_unique_carrier'].unique()
```

```python
flight_df['crs_dep_time'] = flight_df['crs_dep_time'].astype(str).str.zfill(4)
flight_df['crs_dep_time'] = flight_df['crs_dep_time'].str[:2] + ':' + flight_df['crs_dep_time'].str[2:]
```

```python
flight_df['dep_time'] = flight_df['dep_time'].astype(str).str.zfill(4)
flight_df['dep_time'] = flight_df['dep_time'].str[:2] + ':' + flight_df['dep_time'].str[2:]

flight_df['crs_arr_time'] = flight_df['crs_arr_time'].astype(str).str.zfill(4)
flight_df['crs_arr_time'] = flight_df['crs_arr_time'].str[:2] + ':' + flight_df['crs_arr_time'].str[2:]

flight_df['arr_time'] = flight_df['arr_time'].astype(str).str.zfill(4)
flight_df['arr_time'] = flight_df['arr_time'].str[:2] + ':' + flight_df['arr_time'].str[2:]
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
flight_df['origin'].value_counts().head(10).plot(kind='bar')

plt.title('Top 10 Busiest Origin Airports')
plt.xlabel('Airport')
plt.ylabel('Number of Departures')
plt.show()
```

```python
flight_df.groupby('op_unique_carrier')['dep_delay'].mean().plot(kind='bar', figsize=(10,5))

plt.title('Flight Delays by Airline')
plt.xlabel('Airline')
plt.ylabel('Average Delay (Minutes)')
plt.show()
```

```python
sns.boxplot(x='dep_delay', data=flight_df, color='green')


plt.show()
```

```python
flight_df['month'].value_counts().sort_index().plot(
    kind='line', marker='o', figsize=(10,5)
)

plt.title('Flights per Month')
plt.xlabel('Month')
plt.ylabel('Number of Flights')
plt.grid(True)
plt.show()
```

```python
flight_df.groupby('month')['arr_delay'].mean().plot(
    kind='line', marker='o', figsize=(10,5)
)

plt.title('Average Arrival Delay by Month')
plt.xlabel('Month')
plt.ylabel('Average Delay (Minutes)')
plt.grid(True)
plt.show()
```
